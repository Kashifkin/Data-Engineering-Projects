##Vehicles

#Data Cleaning and Transforming

```python
drivers = drivers.dropDuplicates()

drivers = drivers.fillna({
    "first_name": "unknown",
    "last_name": "unknown",
    "phone_number": "not_provided",
    "vehicle_id": "unknown",
    "driver_rating": 0.0,
    "city": "unknown"
})

drivers = drivers.withColumn("driver_rating", col("driver_rating").cast("double"))

drivers = drivers.withColumn(
    "rating_category",
    when(col("driver_rating") < 3.0, "poor")
    .when((col("driver_rating") >= 3.0) & (col("driver_rating") < 4.0), "average")
    .when((col("driver_rating") >= 4.0) & (col("driver_rating") < 4.5), "good")
    .otherwise("excellent")
)

drivers = drivers.withColumn("Full_Name",concat(col("first_name"),lit(" "),col("last_name")))

drivers.display()
```

```python
customers = customers.dropDuplicates()

customers = customers.fillna({
    "customer_id": 0,
    "first_name": "unknown",
    "last_name": "unknown",
    "email": "not_provided",
    "phone_number": "not_provided",
    "city": "unknown"
})

customers = customers.withColumn("customer_id", col("customer_id").cast("int"))

customers = customers.withColumn(
    "signup_date",
    to_date(col("signup_date"), "yyyy-MM-dd"))

customers = customers.withColumn(
    "full_name",concat_ws(" ", col("first_name"), col("last_name")))

customers = customers.withColumn(
    "last_updated_timestamp",
    to_timestamp(col("last_updated_timestamp")))

display(customers)
```

```python
vehicles = vehicles.dropDuplicates()

vehicles = vehicles.fillna({
    "license_plate": "unknown",
    "model": "unknown",
    "make": "unknown",
    "year": 0,
    "vehicle_type": "unknown"
})

vehicles = vehicles.withColumn("year", col("year").cast("int"))
vehicles = vehicles.withColumn("Vehicle_age",2026 - col("year"))
vehicles = vehicles.withColumn("Vehicle_Category", when(col("vehicle_type") == "Sedan" , "Standard")
                               .when(col("vehicle_type") == "Hatchback" , "Economy")
                               .when(col("vehicle_type") == "SUV" , "Premium")
                               .when(col("vehicle_type") == "Luxury" , "Luxury")
                               .when(col("vehicle_type") == "Van" ,"commercial"))

display(vehicles)
```

```python
trips = trips.dropDuplicates()

trips = trips.fillna({
    "start_location": "unknown",
    "end_location": "unknown",
    "distance_km": 0.0,
    "fare_amount": 0.0,
    "payment_method": "unknown",
    "trip_status": "unknown"
})

trips = trips.withColumn("distance_km", col("distance_km").cast("double")) \
                   .withColumn("fare_amount", col("fare_amount").cast("double"))


trips = trips.withColumn("Trips_Distance",when(col('distance_km') < 5 ,"Short" )
                         .when(col('distance_km') > 5-15 ,"medium")
                         .when(col('distance_km') > 15-30 ,"long")
                         .when(col('distance_km') > 30 ,"very_long"))
display(trips)
```

```python
locations = locations.dropDuplicates()

locations = locations.fillna({
    "city": "unknown",
    "state": "unknown",
    "country": "unknown",
    "latitude": 0.0,
    "longitude": 0.0
})

locations = locations.withColumn("latitude", col("latitude").cast("double")) \
                           .withColumn("longitude", col("longitude").cast("double"))

locations = locations.withColumn("hemisphere",when(col("latitude") >= 0, 
                                     "North").otherwise("South")
)

display(locations)
```

```python
payments = payments.dropDuplicates()

payments = payments.fillna({
    "payment_method": "unknown",
    "payment_status": "pending",
    "amount": 0.0
})

payments = payments.withColumn("amount", col("amount").cast("double"))

display(payments)
```

##Gold Tables(Business View)

```python
customer_trips = customers.join(trips, customers.customer_id == trips.customer_id, "inner") \
                             .groupBy(customers.customer_id, customers.full_name) \
                             .agg(count("trip_id").alias("total_trips"))

customer_trips.display()
```

```python
vehicle_trips = trips.join(vehicles, trips.vehicle_id == vehicles.vehicle_id, "inner") \
                     .groupBy(vehicles.vehicle_type) \
                     .agg(count("trip_id").alias("total_trips"))

vehicle_trips.display()
```

```python
payment_summary = payments.groupBy("payment_method") \
                          .agg(round(sum("amount"),2).alias("total_amount"))

payment_summary.display()
```

```python
driver_earnings = trips.join(drivers, trips.driver_id == drivers.driver_id, "inner") \
                          .groupBy(drivers.driver_id, drivers.Full_Name) \
                          .agg(round(sum(trips.fare_amount),2).alias("total_earnings"))

driver_earnings.display()
```

```python


customer_trip_join = customers.join(trips, customers.customer_id == trips.customer_id, "left") \
                             .join(payments, trips.trip_id == payments.trip_id, "left") \
                             .join(drivers, trips.driver_id == drivers.driver_id, "left") \
                             .join(vehicles, trips.vehicle_id == vehicles.vehicle_id, "left") \
                             .join(locations, (trips.start_location == locations.city) | (trips.end_location == locations.city), "left")


customer_summary = customer_trip_join.groupBy(customers.customer_id, 
                                              customers.full_name, 
                                              customers.email, 
                                              customers.city, 
                                              drivers.Full_Name.alias("driver_name"), 
                                              vehicles.vehicle_type) \
                                     .agg(
                                         count(trips.trip_id).alias("total_trips"),
                                         round(sum(payments.amount), 2).alias("total_payment")
                                     )


customer_summary_filtered = customer_summary.filter(col("total_trips") > 3) \
                                            .orderBy(col("total_payment").desc())


customer_summary_filtered.display()
```

```python
customer_summary_filtered.write \
    .mode("overwrite") \
    .saveAsTable("gold_customer_summary")
```
